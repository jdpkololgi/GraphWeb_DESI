## LOA BGS magnitude-limited catalog — interactive subsample

This notebook visualizes a random subsample of the LOA BGS *magnitude-limited* catalog you built from `zall-pix-loa.fits` using only:

- `ZWARN == 0`
- `DELTACHI2 >= 25`
- `BGS_TARGET != 0`

It renders:

- an interactive **3D comoving** scatter (Mpc/h) using Plotly
- an interactive **RA/DEC** sky scatter colored by redshift


```python
import numpy as np
import pandas as pd
import fitsio

import plotly.express as px

from astropy.cosmology import Planck18 as cosmo
```

```python
CATALOG_PATH = "/pscratch/sd/d/dkololgi/graphweb_desi/catalogs/bgs_maglim_zall_zwarn0_dchi2ge25.fits"
N_SUBSAMPLE = 200_000  # adjust for responsiveness
SEED = 42

# Column names in the catalog (from zall ZCATALOG)
COL_RA = "TARGET_RA"
COL_DEC = "TARGET_DEC"
COL_Z = "Z"

COLS = ["TARGETID", COL_RA, COL_DEC, COL_Z, "ZWARN", "DELTACHI2", "BGS_TARGET", "SPECTYPE"]
```

```python
with fitsio.FITS(CATALOG_PATH) as f:
    h = f[1]
    n = h.get_nrows()

print("rows:", n)
print("subsample:", min(N_SUBSAMPLE, n))
```

```python
rng = np.random.default_rng(SEED)
idx = rng.choice(n, size=min(N_SUBSAMPLE, n), replace=False)
idx.sort()

with fitsio.FITS(CATALOG_PATH) as f:
    tab = f[1].read(rows=idx.tolist(), columns=COLS)

df = pd.DataFrame({c: tab[c] for c in tab.dtype.names})

# Basic cleaning
for c in [COL_RA, COL_DEC, COL_Z]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df = df[np.isfinite(df[COL_RA]) & np.isfinite(df[COL_DEC]) & np.isfinite(df[COL_Z])].copy()

print("df shape:", df.shape)
df.head()
```

```python
# Convert to comoving Cartesian coordinates (Mpc/h)
d = cosmo.comoving_distance(df[COL_Z].to_numpy()).value * float(cosmo.h)  # Mpc/h
ra = np.deg2rad(df[COL_RA].to_numpy())
dec = np.deg2rad(df[COL_DEC].to_numpy())

x = d * np.cos(dec) * np.cos(ra)
y = d * np.cos(dec) * np.sin(ra)
z = d * np.sin(dec)

df["X_MPC_H"] = x
df["Y_MPC_H"] = y
df["Z_MPC_H"] = z

print(df[[COL_Z, "X_MPC_H", "Y_MPC_H", "Z_MPC_H"]].describe().loc[["min", "max"]])
```

```python
# Interactive sky view (RA/DEC colored by redshift)
fig_sky = px.scatter(
    df,
    x=COL_RA,
    y=COL_DEC,
    color=COL_Z,
    color_continuous_scale="Turbo",
    opacity=0.6,
    render_mode="webgl",
    title=f"LOA BGS mag-limited subsample (N={len(df):,})",
)
fig_sky.update_traces(marker=dict(size=2))
fig_sky.update_layout(height=600)
fig_sky.show()
```

```python
# Interactive 3D comoving view (Mpc/h)
fig_3d = px.scatter_3d(
    df,
    x="X_MPC_H",
    y="Y_MPC_H",
    z="Z_MPC_H",
    color=COL_Z,
    color_continuous_scale="Turbo",
    opacity=0.35,
    title=f"Comoving 3D (Planck18) (N={len(df):,})",
)
fig_3d.update_traces(marker=dict(size=1))
fig_3d.update_layout(height=750)
fig_3d.show()
```
